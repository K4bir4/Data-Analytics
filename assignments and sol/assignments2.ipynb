```python
#sol13

number = int(input(" enter a number:"))
sum = 0
while number > 0:
    digit = number % 10
    sum += digit
    number= number // 10
print(f"the sum of digit is {sum}.")    
```

```python
#prime no check'
number = int(input("enter the num :"))
is_prime = True
if number <=1 :    
    is_prime = False
else:
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            is_prime = False
            break
if is_prime:
    print(f" {number} is prime number.")
else :
    print(f" {number} is not prime number.")           
```
